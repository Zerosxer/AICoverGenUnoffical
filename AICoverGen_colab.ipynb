# AICoverGen Unofficial Edition

Run the cells from top to bottom. The final cell starts the WebUI and prints a temporary public Gradio URL. Keep this Colab session running while using the app.

```python
#@title Clone AICoverGen Unofficial Edition
import os
import subprocess
subprocess.run(['git', 'clone', 'https://github.com/Zerosxer/AICoverGenUnoffical.git'], check=True)
os.chdir('AICoverGenUnoffical')
print('Repository cloned.')
```

```python
#@title Install compatible requirements
import subprocess
subprocess.run(['python', '-m', 'pip', 'install', '-q', 'uv'], check=True)
subprocess.run(['uv', 'python', 'install', '3.10'], check=True)
subprocess.run(['uv', 'venv', '.venv', '--python', '3.10'], check=True)
python = '.venv/bin/python'
subprocess.run([python, '-m', 'pip', 'install', 'pip<24.1', 'Cython==0.29.37'], check=True)
subprocess.run([python, '-m', 'pip', 'install', 'torch==2.5.1', 'torchaudio==2.5.1'], check=True)
from pathlib import Path
requirements = Path('requirements.txt').read_text().splitlines()
requirements = [
    line for line in requirements
    if line.strip()
    and not line.startswith('--extra-index-url')
    and not line.startswith(('torch==', 'torchaudio==', 'pedalboard=='))
]
requirements.append('pedalboard==0.9.23')
Path('/tmp/aicovergen-colab-requirements.txt').write_text('\n'.join(requirements) + '\n')
subprocess.run(['uv', 'pip', 'install', '--python', python, '-r', '/tmp/aicovergen-colab-requirements.txt'], check=True)
subprocess.run(['apt-get', 'update', '-qq'], check=True)
subprocess.run(['apt-get', 'install', '-y', '-qq', 'sox', 'ffmpeg'], check=True)
print('Dependencies installed.')
```

```python
#@title Download required vocal separation and RVC models
import subprocess
subprocess.run(['.venv/bin/python', 'src/download_models.py'], check=True)
```

```python
#@title Run AICoverGen Unofficial Edition WebUI
import subprocess
subprocess.run(['.venv/bin/python', '-u', 'src/webui.py', '--share'], check=True)
```
